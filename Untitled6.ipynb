```python
# convert dates
fmt = '%Y %b %d %I:%M:%S %p'   # e.g. '2021 Apr 11 12:00:00 AM'
df['DATE OCC'] = pd.to_datetime(df['DATE OCC'], format=fmt)
df['Date Rptd'] = pd.to_datetime(df['Date Rptd'], format=fmt)

# drop columns that are almost empty
df = df.drop(columns=['Crm Cd 2', 'Crm Cd 3', 'Crm Cd 4', 'Cross Street'])

# missing weapon = no weapon used
df['Weapon Desc'] = df['Weapon Desc'].fillna('NO WEAPON')
df['Weapon Used Cd'] = df['Weapon Used Cd'].fillna(0)

# missing sex/descent = no individual victim (e.g. vehicle theft)
df['Vict Sex'] = df['Vict Sex'].replace(['H', '-'], 'X').fillna('N/A')
df['Vict Descent'] = df['Vict Descent'].replace('-', 'X').fillna('N/A')

df['Premis Desc'] = df['Premis Desc'].fillna('UNKNOWN')
df['Mocodes'] = df['Mocodes'].fillna('')

# age 0 or negative is not a real age -> treat as missing, fill with median
df.loc[df['Vict Age'] <= 0, 'Vict Age'] = np.nan
median_age = df['Vict Age'].median()
df['Vict Age'] = df['Vict Age'].fillna(median_age)
print("Median age used:", median_age)

# lat/lon 0 -> fill with the median of that area
df.loc[df['LAT'] == 0, ['LAT', 'LON']] = np.nan
df['LAT'] = df['LAT'].fillna(df.groupby('AREA')['LAT'].transform('median'))
df['LON'] = df['LON'].fillna(df.groupby('AREA')['LON'].transform('median'))

df.isnull().sum()
```

```python
# new useful columns
df['Year'] = df['DATE OCC'].dt.year
df['Month'] = df['DATE OCC'].dt.month
df['Hour'] = df['TIME OCC'] // 100
df['Report Delay'] = (df['Date Rptd'] - df['DATE OCC']).dt.days
df[['DATE OCC', 'Year', 'Month', 'Hour', 'Report Delay']].head()
```

```python
age = df['Vict Age']

# Min-Max
df['Age_minmax'] = (age - age.min()) / (age.max() - age.min())

# Z-score
df['Age_zscore'] = (age - age.mean()) / age.std()

# Decimal scaling (max age is 99 -> divide by 100)
df['Age_decimal'] = age / 100

df[['Vict Age', 'Age_minmax', 'Age_zscore', 'Age_decimal']].describe()
```

```python
# equal width
df['Age_eq_width'] = pd.cut(df['Vict Age'], bins=5)
print(df['Age_eq_width'].value_counts().sort_index())
```

```python
# equal frequency
df['Age_eq_freq'] = pd.qcut(df['Vict Age'], q=5, duplicates='drop')   # many ages = median, so some edges repeat
print(df['Age_eq_freq'].value_counts().sort_index())
```

```python
# custom bins based on the histogram
df['Age Group'] = pd.cut(df['Vict Age'], bins=[0, 17, 30, 45, 60, 100],
                         labels=['Minor', '18-30', '31-45', '46-60', '60+'])

df['Time of Day'] = pd.cut(df['Hour'], bins=[-1, 5, 11, 17, 23],
                           labels=['Night', 'Morning', 'Afternoon', 'Evening'])

df['Age Group'].value_counts().plot(kind='bar', title='Victim Age Group')
plt.show()
df['Time of Day'].value_counts().plot(kind='bar', title='Time of Day')
plt.show()
```

```python
# Weighted mean: mean age of each area, weighted by the number of crimes in that area
per_area = df[df['Age_missing'] == 0].groupby('AREA NAME')['Vict Age'].agg(['mean', 'count'])
weighted_mean = (per_area['mean'] * per_area['count']).sum() / per_area['count'].sum()

print("Simple average of the 21 area means:", per_area['mean'].mean())
print("Weighted mean                      :", weighted_mean)
per_area.sort_values('mean')
```

```python
bins = list(range(0, 130, 10))          # class intervals 0-10, 10-20, ...
freq = pd.cut(age, bins=bins, right=False).value_counts().sort_index()

table = pd.DataFrame({'f': freq.values}, index=freq.index)
table['midpoint'] = [i.left + 5 for i in table.index]
table['f*m'] = table['f'] * table['midpoint']
table['cum f'] = table['f'].cumsum()
table
```

```python
N = table['f'].sum()
h = 10

# grouped mean = sum(f*m) / N
g_mean = table['f*m'].sum() / N

# grouped median = L + ((N/2 - CF) / f) * h
pos = (table['cum f'] >= N / 2).values.argmax()      # median class
L  = table.index[pos].left
f  = table['f'].iloc[pos]
CF = table['cum f'].iloc[pos - 1]
g_median = L + ((N / 2 - CF) / f) * h

# grouped mode = L + ((f1 - f0) / (2*f1 - f0 - f2)) * h
pos = table['f'].values.argmax()                     # modal class
L  = table.index[pos].left
f1 = table['f'].iloc[pos]
f0 = table['f'].iloc[pos - 1]
f2 = table['f'].iloc[pos + 1]
g_mode = L + ((f1 - f0) / (2 * f1 - f0 - f2)) * h

print("Grouped mean  :", g_mean)
print("Grouped median:", g_median)
print("Grouped mode  :", g_mode)
```

```python
def dispersion(x):
    q1, q3 = x.quantile(0.25), x.quantile(0.75)
    return pd.Series({
        'Range': x.max() - x.min(),
        'Variance': x.var(),
        'Std Dev': x.std(),
        'RMSD': np.sqrt(((x - x.mean()) ** 2).mean()),        # root mean square deviation from the mean
        'Coeff of Variation (%)': x.std() / x.mean() * 100,
        'Quartile Coeff of Dispersion': (q3 - q1) / (q3 + q1),
    })

pd.DataFrame({'Vict Age': dispersion(age), 'Report Delay (days)': dispersion(delay)})
```

```python
def shape(x):
    q1, q2, q3 = x.quantile([0.25, 0.5, 0.75])
    return pd.Series({
        "Pearson's skewness": 3 * (x.mean() - x.median()) / x.std(),
        "Bowley's skewness": (q3 + q1 - 2 * q2) / (q3 - q1),
        "Skewness (scipy)": stats.skew(x),
        "Kurtosis (excess)": stats.kurtosis(x),
    })

pd.DataFrame({'Vict Age': shape(age), 'Report Delay (days)': shape(delay)})
```

```python
def iqr_outliers(x):
    q1, q3 = x.quantile(0.25), x.quantile(0.75)
    iqr = q3 - q1
    low, high = q1 - 1.5 * iqr, q3 + 1.5 * iqr
    out = x[(x < low) | (x > high)]
    print(f"Q1={q1}  Q3={q3}  IQR={iqr}  fences=({low}, {high})  outliers={len(out)} ({len(out) / len(x) * 100:.1f}%)")
    return out

print("Victim Age  :"); age_out = iqr_outliers(age)
print("Report Delay:"); delay_out = iqr_outliers(delay)
```

```python
from scipy.spatial import distance

# numeric: distance between two crime records (scaled to 0-1 first)
cols = ['Vict Age', 'Hour', 'Report Delay']
X = (df[cols] - df[cols].min()) / (df[cols].max() - df[cols].min())
a, b = X.iloc[0], X.iloc[1]

print(df.loc[[0, 1], ['Crm Cd Desc'] + cols], "\n")
print("Euclidean    :", distance.euclidean(a, b))
print("Manhattan    :", distance.cityblock(a, b))
print("Minkowski p=3:", distance.minkowski(a, b, p=3))
inv_cov = np.linalg.inv(np.cov(X.T))
print("Mahalanobis  :", distance.mahalanobis(a, b, inv_cov))
```

```python
# binary: do two crime types occur at the same kinds of premises?  (1 = that crime has occurred at that premise)
prem = (pd.crosstab(df['Crm Cd Desc'], df['Premis Desc']) > 0).astype(int)
v1 = prem.loc['BURGLARY']
v2 = prem.loc['BURGLARY FROM VEHICLE']

f11 = ((v1 == 1) & (v2 == 1)).sum()
f00 = ((v1 == 0) & (v2 == 0)).sum()
f10 = ((v1 == 1) & (v2 == 0)).sum()
f01 = ((v1 == 0) & (v2 == 1)).sum()
print("f11 =", f11, " f00 =", f00, " f10 =", f10, " f01 =", f01)
print("SMC    :", (f11 + f00) / (f11 + f00 + f10 + f01))
print("Jaccard:", f11 / (f11 + f10 + f01))
```
